```python
import pandas as pd

df = pd.read_csv("data_B.csv")

df.head()
```

```python
df.info()
```

```python
df = df.drop(columns = [
    "Unnamed: 0",
    "ID",
    "Customer_ID",
    "Name",
    "SSN"
])

df.info()
```

```python
import numpy as np

cols_to_numeric = [
    "Age",
    "Annual_Income",
    "Num_of_Loan",
    "Num_of_Delayed_Payment",
    "Changed_Credit_Limit",
    "Outstanding_Debt",
    "Amount_invested_monthly",
    "Monthly_Balance"
]

for col in cols_to_numeric:
    df[col] = pd.to_numeric(df[col], errors = "coerce")

df.info()
```

```python
num_cols = [
    "Age",
    "Annual_Income",
    "Monthly_Inhand_Salary",
    "Num_of_Loan",
    "Num_of_Delayed_Payment",
    "Changed_Credit_Limit",
    "Num_Credit_Inquiries",
    "Outstanding_Debt",
    "Amount_invested_monthly",
    "Monthly_Balance"
]

import matplotlib.pyplot as plt

boxplot = df.boxplot(column = num_cols)

plt.show()
```

```python
for col in num_cols:
    df[col] = df[col].fillna(np.nanmedian(df[col]))

df.info()
```

```python
def convert_credit_history(x):
    if pd.isna(x):
        return np.nan
    
    parts = x.split(" ")
    years = int(parts[0])
    months = int(parts[3])

    return (years * 12) + months

df["Credit_History_Age"] = df["Credit_History_Age"].apply(convert_credit_history)

df["Credit_History_Age"]
```

```python
boxplot = df.boxplot(column = ["Credit_History_Age"])

plt.show()
```

```python
df["Credit_History_Age"] = df["Credit_History_Age"].fillna(np.nanmean(df["Credit_History_Age"]))

df.info()
```

```python
df["Credit_Score"].value_counts()
```

```python
mapping = {
    "Poor": 0,
    "Standard": 1,
    "Good": 2
}
df["Credit_Score"] = df["Credit_Score"].map(mapping)

df["Credit_Score"].value_counts()
```

```python
def clean_loan(x):
    # Case 1: list → clean inside
    if isinstance(x, list):
        x = [str(i).strip() for i in x if str(i).strip().lower() != "not specified"]
        x = [i for i in x if i != ""]
        return x if len(x) > 0 else np.nan

    # Case 2: normal value → force string
    if pd.isna(x):
        return np.nan

    x = str(x).strip()

    # Remove weird unicode spaces
    x = x.replace("\xa0", "").strip()

    # Empty or whitespace
    if x == "":
        return np.nan

    # Not Specified
    if x.lower() == "not specified":
        return np.nan

    return x

df["Type_of_Loan"] = df["Type_of_Loan"].apply(clean_loan)
mode_val = df["Type_of_Loan"].mode()

mode_val = mode_val.iloc[0]
if isinstance(mode_val, list):
    mode_val = mode_val[0]

df["Type_of_Loan"] = df["Type_of_Loan"].fillna(mode_val)
df["Type_of_Loan"] = df["Type_of_Loan"].replace(r"^\s*$", np.nan, regex=True)
df["Type_of_Loan"] = df["Type_of_Loan"].replace(
    to_replace=r"(?i)not specified",
    value=np.nan,
    regex=True
)
df["Type_of_Loan"] = df["Type_of_Loan"].fillna(
    df["Type_of_Loan"].mode()[0]
)
df["Type_of_Loan"] = df["Type_of_Loan"].fillna("")
df["Type_of_Loan"] = df["Type_of_Loan"].replace(["Not Specified", "_"], "")
df["Type_of_Loan"] = df["Type_of_Loan"].str.replace(" and ", ", ", regex = False)
df["Type_of_Loan"] = df["Type_of_Loan"].apply(
    lambda x: [loan.strip() for loan in x.split(",")] if x != "" else []
)

df["Type_of_Loan"].explode().value_counts()
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
loan_dummies = pd.DataFrame(
    mlb.fit_transform(df["Type_of_Loan"]),
    columns = mlb.classes_,
    index = df.index
)
df = pd.concat([df, loan_dummies], axis = 1)
df = df.drop("Type_of_Loan", axis = 1)
df["num_loan_types"] = loan_dummies.sum(axis = 1)
df["loan_diversity"] = (loan_dummies.sum(axis = 1) > 2).astype(int)

df.info()
```

```python
df["Occupation"].value_counts()
```

```python
df["Occupation"] = df["Occupation"].replace("_______", np.nan)
df["Occupation"] = df["Occupation"].fillna(df["Occupation"].mode()[0])

df["Occupation"].value_counts()
```

```python
df["Credit_Mix"].value_counts()
```

```python
df["Credit_Mix"] = df["Credit_Mix"].replace("_", np.nan)
df["Credit_Mix"] = df["Credit_Mix"].fillna(df["Credit_Mix"].mode()[0])
credit_mix_map = {
    "Bad": 0,
    "Standard": 1,
    "Good": 2
}
df["Credit_Mix"] = df["Credit_Mix"].map(credit_mix_map)

df["Credit_Mix"].value_counts()
```

```python
df["Payment_of_Min_Amount"].value_counts()
```

```python
df["Payment_of_Min_Amount"] = df["Payment_of_Min_Amount"].replace("NM", np.nan)
df["Payment_of_Min_Amount"] = df["Payment_of_Min_Amount"].fillna(df["Payment_of_Min_Amount"].mode()[0])
payment_of_min_amount_map = {
    "No": 0,
    "Yes": 1
}
df["Payment_of_Min_Amount"] = df["Payment_of_Min_Amount"].map(payment_of_min_amount_map)

df["Payment_of_Min_Amount"].value_counts()
```

```python
df["Payment_Behaviour"].value_counts()
```

```python
df["Payment_Behaviour"] = df["Payment_Behaviour"].replace("!@9#%8", np.nan)
df["Payment_Behaviour"] = df["Payment_Behaviour"].fillna(df["Payment_Behaviour"].mode()[0])
df["spending_level"] = df["Payment_Behaviour"].apply(lambda x: x.split("_")[0])
df["payment_size"] = df["Payment_Behaviour"].apply(lambda x: x.split("_")[2])
df["spending_level"] = df["spending_level"].map({
    "Low": 0,
    "High": 1
})
payment_size_map = {
    "Small": 0,
    "Medium": 1,
    "Large": 2
}
df["payment_size"] = df["payment_size"].map(payment_size_map)
df = df.drop("Payment_Behaviour", axis = 1)

df["spending_level"].value_counts()
```

```python
df["payment_size"].value_counts()
```

```python
cat_cols = [
    "Month",
    "Occupation"
]

for col in cat_cols:
    df = pd.get_dummies(df, columns = [col], prefix = col, dtype = float)

df.info()
```

```python
df["debt_to_income"] = df["Outstanding_Debt"] / df["Annual_Income"].replace(0, np.nan)

df.info()
```

```python
df["debt_per_loan"] = df["Outstanding_Debt"] / df["Num_of_Loan"].replace(0, np.nan)

df.info()
```

```python
boxplot = df.boxplot(column = ["debt_per_loan"])

plt.show()
```

```python
df["debt_per_loan"] = df["debt_per_loan"].fillna(np.nanmedian(df["debt_per_loan"]))

df.info()
```

```python
df["delay_ratio"] = df["Num_of_Delayed_Payment"] / df["Num_of_Loan"].replace(0, np.nan)

df.info()
```

```python
boxplot = df.boxplot(column = ["delay_ratio"])

plt.show()
```

```python
df["delay_ratio"] = df["delay_ratio"].fillna(np.nanmedian(df["delay_ratio"]))

df.info()
```

```python
df["high_utilization"] = (df["Credit_Utilization_Ratio"] > 0.3).astype(int)

df.info()
```

```python
df["age_group"] = pd.cut(
    df["Age"],
    bins = [18, 25, 35, 50, 100],
    labels = ["young", "adult", "mid", "senior"]
)
df = pd.get_dummies(df, columns = ["age_group"], dtype = float)

df.info()
```

```python
df["long_credit_history"] = (df["Credit_History_Age"] > 120).astype(int)

df.info()
```

```python
df["income_delay"] = df["Annual_Income"] / df["Num_of_Delayed_Payment"].replace(0, np.nan)

df.info()
```

```python
boxplot = df.boxplot(column = ["income_delay"])

plt.show()
```

```python
df["income_delay"] = df["income_delay"].fillna(np.nanmedian(df["income_delay"]))

df.info()
```

```python
df["credit_history_per_age"] = df["Credit_History_Age"] / df["Age"]

df.info()
```

```python
df["total_credit_lines"] = df["Num_Bank_Accounts"] + df["Num_Credit_Card"]

df.info()
```

```python
corr = df.corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape), k = 1).astype(bool))
to_drop = [column for column in upper.columns if any(upper[column] > .95)]
df = df.drop(columns = to_drop)

df.info()
```

```python
for col in df.select_dtypes(include = ["float64", "int64"]).columns:
    q1 = df[col].quantile(0.01)
    q99 = df[col].quantile(0.99)
    df[col] = df[col].clip(q1,q99)

df.info()
```

```python
df = df.drop(columns = [df.columns[20]])

df.info()
```

```python
from sklearn.model_selection import train_test_split

x = df.drop("Credit_Score", axis = 1)
y = df["Credit_Score"]
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size = 0.2, random_state = 42, stratify = y
)

x_train.info()
```

```python
x_test.info()
```

```python
y_train.info()
```

```python
y_test.info()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler

rf = RandomForestClassifier()

scaler = StandardScaler()
scaler_train = scaler.fit_transform(x_train)
scaler_test = scaler.transform(x_test)

rf.fit(scaler_train, y_train)

y_pred = rf.predict(scaler_test)

print("Base Classification Report\n")
print(classification_report(y_test, y_pred, target_names = ["Standard", "Poor", "Good"]))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [200, 300, 500],
    "max_depth": [10, 15, 20, None],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2"]
}
grid = GridSearchCV(rf, param_grid, cv = 5, scoring = "accuracy", n_jobs = -1)
```

```python
grid.fit(scaler_train, y_train)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best Score:", grid.best_score_)
```

```python
best = RandomForestClassifier(
    max_depth = None, 
    max_features = "sqrt", 
    min_samples_leaf = 1, 
    n_estimators = 200
)

best.fit(scaler_train, y_train)

y_pred_best = best.predict(scaler_test)

print("Best Classification Report\n")
print(classification_report(y_test, y_pred_best, target_names = ["Standard", "Poor", "Good"]))
```

```python
import pickle as pkl
filename = 'rf_best_class.pkl'
pkl.dump(best, open(filename, 'wb'))
```
